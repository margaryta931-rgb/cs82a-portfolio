```python
import seaborn as sns
import matplotlib.pyplot as plt

# Check missing values
print("Missing values:")
print(df.isnull().sum())

# Find average flipper length for each species
avg_flipper = df.groupby("species")["flipper_length_mm"].mean()

print("\nAverage flipper length:")
print(avg_flipper)

# Make a chart
sns.barplot(data=df, x="species", y="flipper_length_mm")

plt.title("Average Flipper Length by Species")
plt.xlabel("Penguin Species")
plt.ylabel("Average Flipper Length (mm)")
plt.show()
```
