```python
import sqlite3
import pandas as pd

con = sqlite3.connect("module6_lab2.db")
cur = con.cursor()

print("Database connected")
```

```python
cur.execute("""
CREATE TABLE IF NOT EXISTS customers (
    id INTEGER PRIMARY KEY,
    name TEXT,
    city TEXT,
    total_spent REAL
)
""")

con.commit()

print("Customers table created")
```

### MySQL to SQLite Note

SQLite uses a simpler data type system than MySQL. I used INTEGER, TEXT, and REAL data types when creating the customers table.

```python
customers = [
    (1, "Olivia", "San Diego", 125.50),
    (2, "Ethan", "Sacramento", 210.75),
    (3, "Chloe", "San Francisco", 89.99),
    (4, "Noah", "Fresno", 340.20),
    (5, "Grace", "San Jose", 175.00),
    (6, "Lucas", "Riverside", 95.40),
    (7, "Mia", "Anaheim", 260.30),
    (8, "Henry", "Santa Barbara", 145.80)
]

cur.executemany("""
INSERT OR REPLACE INTO customers (id, name, city, total_spent)
VALUES (?, ?, ?, ?)
""", customers)

con.commit()

print("8 customers inserted")
```

```python
# Query 1: WHERE
query1 = """
SELECT *
FROM customers
WHERE total_spent > 150;
"""

result1 = pd.read_sql_query(query1, con)
result1
```

```python
# Query 2: ORDER BY
query2 = """
SELECT *
FROM customers
ORDER BY total_spent DESC;
"""

result2 = pd.read_sql_query(query2, con)
result2
```

```python
# Query 3: LIKE
query3 = """
SELECT *
FROM customers
WHERE name LIKE '%a%';
"""

result3 = pd.read_sql_query(query3, con)
result3
```
