```python
import pandas as pd
from sklearn.linear_model import LinearRegression

df = pd.read_csv("ratings.tsv", sep="\t")
df.head()
```

```python
X = df[["ambience", "food", "service"]]
y = df["rating"]

model = LinearRegression()
model.fit(X, y)

print("Coefficients:")
for name, coef in zip(X.columns, model.coef_):
    print(name, ":", round(coef, 4))
```

```python
coefficients = pd.DataFrame({
    "Factor": X.columns,
    "Coefficient": model.coef_
})

coefficients = coefficients.sort_values(
    by="Coefficient", ascending=False
)

print(coefficients.to_string(index=False))
```

I think the restaurant should improve its food first. Food has the highest coefficient (0.7338), so it has the strongest effect on the overall rating in this model. Service is second (0.3240), and ambience is last (0.1338).
