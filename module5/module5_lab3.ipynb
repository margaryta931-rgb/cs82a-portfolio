```python
before_corr = df[['age', 'income', 'num.vehicles']].corr()
before_corr
```

```python
df[['age', 'income', 'num.vehicles']].describe()
```

```python
invalid_rows = df[
    (df['age'] <= 0) |
    (df['age'] > 100) |
    (df['income'] < 0)
]

invalid_rows[['age', 'income', 'num.vehicles']]
```

```python
clean_df = df[
    (df['age'] > 0) &
    (df['age'] <= 100) &
    (df['income'] >= 0)
].copy()

print("Rows before cleaning:", len(df))
print("Rows after cleaning:", len(clean_df))
print("Rows removed:", len(df) - len(clean_df))
```

```python
after_corr = clean_df[['age', 'income', 'num.vehicles']].corr()
after_corr
```

### Data Cleaning

I found 12 rows with incorrect values and removed them from the dataset. The invalid data included ages of 0 or above 100 and one negative income value. These rows were excluded so they would not influence the final correlation results.

### Interpretation

The cleaned results show very little connection between age and income. The number of vehicles slightly decreases as age increases, while people with higher incomes tend to have slightly more vehicles. These correlations are weak, so there is no strong relationship between the variables.
