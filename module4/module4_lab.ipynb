```python
import pandas as pd

df = pd.read_csv("sales_clean.csv")
print(df.shape)
df.head()
```

```python
print("Price mean:", df["price"].mean())
print("Price median:", df["price"].median())
print("Price std:", df["price"].std())

print("Qty mean:", df["qty"].mean())
print("Qty median:", df["qty"].median())
print("Qty std:", df["qty"].std())
```

I would report the median price because the mean is much higher than the median. This shows that a few high prices are pulling the mean up.

```python
import matplotlib.pyplot as plt

df["price"].plot(kind="hist", bins=20)
plt.title("Price Distribution")
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.show()
```

```python
df["revenue"] = df["price"] * df["qty"]
```

```python
corr = df.corr(numeric_only=True)
corr
```

```python
df.plot.scatter(x="qty", y="revenue")
plt.title("Quantity vs Revenue")
plt.xlabel("Quantity")
plt.ylabel("Revenue")
plt.show()
```

When quantity goes up, revenue usually goes up too, but the connection is not very strong and some unusual values can affect the result.

Mean: (2 + 4 + 4 + 6 + 9) / 5 = 5.0

Median: The middle value of [2, 4, 4, 6, 9] is 4.

```python
values = [2, 4, 4, 6, 9]

print("Mean:", sum(values) / len(values))
print("Median:", pd.Series(values).median())
```
