```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("truck.csv")
df.head()
```

```python
df.columns
```

```python
r = df["X"].corr(df["Y"])
print("Pearson correlation r =", r)
```

```python
plt.scatter(df["X"], df["Y"])
plt.xlabel("List Price (thousands of dollars)")
plt.ylabel("Best Sale Price (thousands of dollars)")
plt.title("Truck List Price vs. Best Sale Price")
plt.show()
```

```python
import numpy as np

m, b = np.polyfit(df["X"], df["Y"], 1)

list_price = 20
predicted_price = m * list_price + b

print("List price:", list_price)
print("Predicted best price:", predicted_price)
```

The predicted sale price for a $20,000 truck is about $17,458. I think this prediction is reliable because the points are very close to a straight line. It may be less accurate for prices outside this data range.
