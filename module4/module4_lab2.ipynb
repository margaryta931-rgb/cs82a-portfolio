```python
print("Mean weight:", books["weight"].mean())
print("Mean weight by cover:")
print(books.groupby("cover")["weight"].mean())

print("Sorted covers:")
print(books["cover"].sort_values().tolist())
```

### Results

1. The median cover type is **pb (paperback)**.
   Paperback is the middle cover type.

2. The mean weight of all books is **706.67 grams**.
   Books weigh around 707 grams on average.

3. The mean weight for hardcover books is **796.43 grams**, while the mean weight for paperback books is **628.13 grams**.
   Hardcover books weigh more than paperback books on average.
